#### Sorting and Filtering Data


```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name":["Rahul" , "Neha" , "Priya" , "Karan" , "Riya"],
    "Age":[25 , 32 , 22 , 24 , 23],
    "Gender":['Male' , 'Female' , 'Female' , 'Male' , 'Female'],
    "Class":[1 , 2 , 1 , 1 , 3],
    "Fare":[500 , 800 , 1200 , 650 , 1800] 
})

print(df)
print(df.dtypes)
```

```python
# sort by numerical column

sorted_df = df.sort_values("Age")

print(sorted_df)
```

```python
# Filter rows where Age > 25

filtered_df = df['Age'].where(df['Age'] > 25)

filter_df = df[df['Age'] <= 25]

print(filtered_df)

print(filter_df)
```

```python
# Aggregation and Grouping

# Group by Gender

gender_df = df.groupby("Gender")

print(gender_df)

summary = gender_df["Fare"].agg(["mean" , "sum" , "count"])

print(summary)
```

```python
# group with higest average fare

highest_average = summary["mean"].idxmin()

print(highest_average)
```

#### Data Visulization

```python
# Bar Chart: Total count for each gender

gernder_count = df["Gender"].value_counts()

plt.figure(facecolor="white")
axis = gernder_count.plot(kind="bar" , color=["lavender" , "lavender"])
axis.set_facecolor("blue")
plt.title("Total Person by Gender")
plt.xlabel("Gender" , color="Red")
plt.ylabel("Count" , color="Green")
plt.show()
```

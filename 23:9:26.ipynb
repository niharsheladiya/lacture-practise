#### Introduction to MATPLOTLIB



matplotlib is a python library used to create charts and visualizations.

Common charts:
1. Bar Chart

2. Line charts

3. Scatter plot

4. Pie chart

5. Histogram



Common Keyword and Functions


```python
# plt.figure() : Create a new figure.
# plt.title() : Add chart title.
# plt.xlabel() : Add X-axis label.
# plt.ylabel() : Add Y-axis label.
# plt.xticks() : Change X-axis tick labels.
# plt.yticks() : Change Y-axis tick labels.
# plt.legend() : Display legend.
# plt.grid() : Display grid lines.
# plt.show() : Display the chart.
# plt.savefig() : Save chart as an image.
# plt.xlim() : Get X-axis limit
# plt.ylim() : Get Y-axis limit
# plt.text() : Add text to chart.
# plt.annotate() : Add annotation to chart.
# aplha : Controls transparancy.
# color : Set color.
# label : Set legend label.
# marker : Set point style.
# linestyle : set line style.
# linewidth : Set line thickness.
# figsize : Set chart width and height.
```

```python
import matplotlib.pyplot as plt

# BAR CHART

products = ["Laptop" , "Mobile" , "Tablet" , "Moniter" , "Keyboard"]

sales = [50 , 80 , 40 , 20 , 60]

plt.figure(figsize=(8 , 5))

plt.bar(products , sales , color="brown" , label="Sales")

plt.title("ProductnSales")

plt.xlabel("Products")

plt.ylabel("Number of Sales")

plt.legend()

plt.show()


        
```

```python
# Line Chart

month = ["JAN" , "FEB" , "MAR" , "APR" , "MAY" , "JUN" , "JUL" , "AUG" , "SEPT" , "OCT" , "NOV" , "DEC"]

temperature = [20 , 20 , 30 , 40 , 50 , 30 , 30 , 35 , 30 , 35 , 27 , 23]

plt.figure(figsize=(10 , 5))

plt.plot(
    month,
    temperature,
    marker="o",
    color="orange",
    linestyle="-",
    linewidth=2,
    label="Temparature"
)

plt.title("Monthly Temperature")

plt.xlabel("Month")

plt.ylabel("Temperature (oc)")

plt.grid(True)

plt.legend()

plt.show()

```

```python
#Scatter chart

temperature = [14.2 , 16.4 , 11.9 , 15.2 , 18.5 , 22.1 , 19.4 , 25.1 , 23.4 , 18.1 , 22.6 , 17.2]

sales = [215 , 325 , 185 , 332 , 406 , 522 , 412 , 614 , 544 , 421 , 445 , 408]

plt.figure(figsize=(8 , 5))

plt.scatter(
    temperature,
    sales,
    color="green",
    marker="o",
    label="Sales study in Temperature"
)

plt.title("Temperature vs Sales Data")

plt.xlabel("Temperature")

plt.ylabel("Sales")

plt.grid(True)

plt.legend()

plt.show()

```

```python
# PIE CHART

fruits = ["Apple" , "Banana" , "Orange" , "Kiwi" , "Grapes"]

fruits_sales = [25 , 20 , 30 , 15 , 10]

plt.figure(figsize=(7 , 7))

plt.pie(fruits_sales, labels=fruits, startangle=90)

plt.title("Fruit Sales Distribution")

plt.show()
```

```python
# HISTOGRAM

ages = [
    12, 15, 17, 19, 22, 24, 25, 25, 26, 27, 28, 29, 31, 32, 34, 35, 
    35, 36, 38, 41, 42, 45, 47, 52, 55, 59, 61, 65, 72, 75
]

plt.figure(figsize=(8 , 5))

plt.hist(
    ages,
    bins=10,
    color="purple",
    edgecolor="black",
    alpha=0.7,
    linewidth=2
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.grid(axis="y")

plt.show()
```

```python
# Merging DataFrame

import pandas as pd


employee_info = pd.DataFrame({
    "Employee_ID":[101 , 102 , 103 , 104],
    "Name":["Rahul" , "Amit" , "Priya" , "Neha"]
})

employee_salary = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104],
    "Salary": [20000, 40000, 30000, 55000]
})

print(employee_info)
print(employee_salary)
```

```python
# inner join

print("Inner join")

inner_df = pd.merge(
    employee_info,
    employee_salary,
    on="Employee_ID",
    how="inner"
)

print(inner_df)
```

```python
# outer join

print("Outer join")

outer_df = pd.merge(
    employee_info,
    employee_salary,
    on="Employee_ID",
    how="outer"
)

print(outer_df)
```

```python
# Merge Parameter

print("Merge Parameter")

print("""
  pd.merge(
    left
    right
    on
    how
  )

  left : First DataFrame

  right : Second DataFrame

  on : common column used for merging

  how : "inner" : only matching rows.

  how : "outer" : All rows from both DataFarme.

  how : "left" : All rows from left DataFrame.

  how : "right" : All rows from right DataFrame.



"""
)

# CROSS JOIN

print("OUTER JOIN")

outer_df = pd.merge(
  employee_info,
  employee_salary,
  how="cross"
)

print(outer_df)
```

```python
# LEFT JOIN

print("OUTER JOIN")

left_df = pd.merge(
  employee_info,
  employee_salary,
  on = "Employee_ID",
  how="left"
)

print(left_df)
```

```python
# RIGHT JOIN

print("OUTER JOIN")

right_df = pd.merge(
  employee_info,
  employee_salary,
  on = "Employee_ID",
  how="right"
)

print(right_df)
```

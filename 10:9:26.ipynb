```python
import pandas as pd
```

```python
# DataFrames

student_data = {
    'name': ['Alice', 'Bob', 'Rahu' , 'Amit' , 'Vivek' , 'Nihar'],
    'age' : [10 , 20 , 30 , 40 , 50 , 60],
    'area' : ['Delhi' , 'Mumbai' , 'Bangalore' , 'Chennai' , 'Kolkata' , 'Pune'],
    'score' : [90 , 80 , 70 , 60 , 50 , 40]
}

student_df = pd.DataFrame(student_data)

print(student_df)
```

#### Access DataFrame values

```python
print("First 2 Rows:\n" , student_df.head(2))

print("Last 2 Rows:\n" , student_df.tail(2))

print("Index : \n" , student_df.index)

print("Data Types: \n" , student_df.dtypes)

print("Columns: \n" , student_df.columns)

print("Info:" , student_df.info())

print("statistical Summary: \n" , student_df.describe())

# Select columns and rows

print("Name Column: \n" , student_df['name'])

print("Score Column: \n" , student_df['score'])

print("Area Column: \n" , student_df['area'])

print("Second Row: \n" , student_df.iloc[2])

print("Second Row :\n" , student_df.loc[0:2])

# DataFrame Operations
# Add a new column

student_df['email'] = ["example1@gmail.com" , "example2@gmail.com" , "example3@gmail.com" , "example4@gmail.com" , "example5@gmail.com" , "example6@gmail.com"]

print(student_df)

student_df = student_df.drop('email' , axis=1)

print(student_df)

# ages = student_df[student_df['age'] == 20] aama error che jya age ni jagya ee aages lakhyu che nich

ages = student_df[student_df['age'] == 20]

soted_df = student_df.sort_values(by="name")

print(ages)

print(soted_df)


# mean()

average_age = student_df['age'].mean() # ahiya pan same evi rite jj

print(average_age)


```

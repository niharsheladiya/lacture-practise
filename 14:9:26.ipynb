### Handling Dates and Times


```python
date_time_df = pd.DataFrame({
    'name':['A' , 'B' , 'C' , 'D'],
    'age':[20 , 25 , 21 , 30],
    'salary':[10000 , 21000 , 40000 , 50000],
    'dob':['2026-09-10' , '2026-10-11' , '2026-11-12' , '2026-12-14']
})

print(date_time_df)

print(date_time_df.dtypes)
```

```python
# date_time_df = convert string into datetime

date_time_df['dob'] = pd.to_datetime(date_time_df['dob'])

print(date_time_df)

print(date_time_df.dtypes)
```

```python
# Extract YEAR , MONTH AND DAY

import datetime as data_types_df

# dt.tear

# Extract YEAR, MONTH, and DAY

date_time_df['year'] = date_time_df['dob'].dt.year

print(date_time_df['year'])

date_time_df['month'] = date_time_df['dob'].dt.month

print(date_time_df['month'])

date_time_df['day'] = date_time_df['dob'].dt.day

print(date_time_df['day'])



# Other properties in datetime module

date_time_df['day_name'] = date_time_df['dob'].dt.day_name()

print(date_time_df['day_name'])

date_time_df['month_name'] = date_time_df['dob'].dt.month_name()

print(date_time_df['month_name'])

```

```python
# Filter Data Using Dtae Range

start_date = '2026-09-01'
end_date = '2026-11-15'

filter_date_df = date_time_df[(date_time_df['dob'] >= start_date) & (date_time_df['dob'] <= end_date)]

print(filter_date_df)
```

```python
# Diffrenece between two Dates

employee_diff_df = pd.DataFrame({
    'employee':['Raj' , 'Dhaval' , 'Manish' , 'Rudra' , 'Digant'],
    'joining_date':['2026-01-01' , '2026-02-10' , '2026-03-01' , '2026-04-01' , '2026-05-01'],
    'leaving_date':['2026-01-20' , '2026-03-01' , '2026-05-01' , '2026-05-15' , '2026-07-07']

})

print(employee_diff_df)

print(employee_diff_df.dtypes)
```

```python
# convert both columns to datetime

employee_diff_df['joining_date'] = pd.to_datetime(employee_diff_df['joining_date'])

employee_diff_df['leaving_date'] = pd.to_datetime(employee_diff_df['leaving_date'])

print(employee_diff_df.dtypes)
```

```python
# Diffrence between dates

employee_diff_df['days'] = (employee_diff_df['leaving_date'] - employee_diff_df['joining_date']).dt.days

print(employee_diff_df['days'])
```

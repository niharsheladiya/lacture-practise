#### Stack Plot

```python
import matplotlib.pyplot as plt
```

```python
# Sales of Product A , B and C over 12 month

month = ["JAN" , "FEB" , "MAR" , "APR" , "MAY" , "JUN" , "JUL" , "AUG" , "SEP" , "OCT" , "NOV" , "DEC"]

product_a = [20 , 25 , 30 , 35 , 40 , 45 , 50 , 55 , 60 , 65 , 70 , 75]

product_b = [15 , 20 , 25 , 30 , 35 , 40 , 45 , 50 , 55 , 60 , 65 , 70]

product_c = [10 , 15 , 20 , 25 , 30 , 35 , 40 , 45 , 50 , 55 , 60 , 65]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    month,
    product_a,
    product_b,
    product_c,
    labels=[
        "Product A",
        "Product B",
        "Product C"
    ],
    colors=[
        "skyblue",
        "orange",
        "lightgreen"
    ]
)

plt.title("Monthly Sales of Product")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.legend()

plt.show()

```

```python
# Stack Plot

# Revenue of HR , Marketing , Sales and IT

# over 6 month

month = ["JAN" , "FEB" , "MAR" , "APR" , "MAY" , "JUN"]

hr = [20 , 26 , 28 , 30 , 32 , 34]

marketing = [35 , 40 , 45 , 50 , 55 , 60]

sales = [10 , 25 , 35 , 55 , 75 , 95]

it = [25 , 20 , 30 , 35 , 45 , 35]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    month,
    hr,
    marketing,
    sales,
    it,
    labels=[
        "HR",
        "Marketing",
        "Sales",
        "IT"
    ],
    colors=[
        "blue",
        "green",
        "black",
        "red"
    ]
)

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.legend()

plt.title("Department Revenue Distribution")

plt.show()
```

#### Subplot

```python
# one bar chart + one line chart

students = ["Rahul" , "Amit" , "Priya" , "Neha" , "Pooja"]

marks = [70 , 80 , 89 , 92 , 67]

months = ["Jan" , "Feb" , "Mar" , "Apr" , "May"]

sales = [100 , 150 , 200 , 250 , 300]

fig , axes = plt.subplots(1 , 2 , figsize=(12 , 5))

axes[0].bar(students , marks , color="skyblue")

axes[0].set_title("Student Marks")

axes[0].set_xlabel("Students")

axes[0].set_ylabel("Marks")

axes[1].plot(
    months,
    sales,
    marker="o",
    color="red"
)

axes[1].set_title("Monthly Sales")

axes[1].set_xlabel("Month")

axes[1].set_ylabel("Sales")

plt.tight_layout()

plt.savefig("Student&SalesCharts.pdf")

plt.show()
```

```python
# PIE Chart

# Market Share of 5 Companies

companies = ["TATA" , "Relience" , "Jio" , "Airtel" , "Vodafone" , "Idea"]

market_share = [30 , 25 , 34 , 35 , 15 , 10]

plt.figure(figsize=(7 , 7))

plt.pie(
  market_share,
  labels=companies,
  autopct="%1.2f%%",
  startangle=90
)

plt.title("Market Share of Companies")

plt.savefig("company_market_share.png" , dpi=600 , bbox_inches="tight")

plt.show()
```

#### Pandas
#### Series / Dataframes / Dataframe Operations

```python
# Series

fruits = pd.Series(['Apple', 'Banana', 'Mango', 'Orange' , 'cherry'])

print(fruits)
```

```python
# Index : custom labels for series

colors = pd.Series(['Red', 'Yellow', 'Green', 'Orange', 'Red'], index=['color1', 'color2', 'color3', 'color4', 'color5'])


result = colors.to_string(index=False)

print(result)
print(colors.values)
print(colors.tolist())

```

```python
# iloc  = access data using integer position

print(colors.iloc[1])

# loc = access data using label

print(colors.loc['color3'])

# len = number of elements

print(len(colors))

# unique = return unique value

print(colors.unique())

print(colors.nunique())

# value_counts() = frequency of each value

print(colors.value_counts())
```

```python
# DATAFRAMES

student_data = {
  'name':['Alice' , 'Bob' , 'Rahul' , 'Amit'],
  'age':[10 , 20 , 30 , 40],
  'area':['surat' , 'vadodara' , 'mumbai' , 'pune']
}

student_df = pd.DataFrame(student_data)

print(student_df)
```

#### Stack Plot

```python
import matplotlib.pyplot as plt
```

```python
# Sales of Product A  ,B and C over 12 month

month = ["JAN" , "FEB" , "MAR" , "APR" , "MAY" , "JUN" , "JUL",  "AUG" , "SEPT" , "OCT" , "NOV" , "DEC"]

product_a = [20 , 25 , 30 , 35 , 40 , 45 , 50 , 55 , 60 , 65 , 70 , 75]

product_b = [15 , 20 , 25 , 30 ,35 , 40 , 45 , 50 , 55 , 60 , 65 , 70 ]

product_c = [10 , 15 , 20 , 25 , 30 ,35 , 40 , 45 , 50 , 55 , 60 , 65]

plt.figure(figsize=(10 , 5))

plt.stackplot(
  month,
  product_a,
  product_b,
  product_c,
  labels=[
    "Product A",
    "Product B",
    "Product C"
  ],
  colors=[
    "skyblue",
    "orange",
    "lightgreen"
  ]
)

plt.title("Monthly Sales of Product")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.legend()

plt.show()
```

```python
# Stack Plot

# Revenue of HR , Marketing  , Sales and IT

# over 6 month

month = ["JAN" , "FEB" , "MAR" , "APR" , "MAY" , "JUN"]

hr = [20 , 26 , 28 , 30 , 32 , 34]
marketing = [35 , 40 , 45 , 50 , 55 , 60]
sales = [10 , 25 , 35 , 55 , 75 , 95]
it = [25 , 20 , 30 , 35 , 45 , 35]


plt.figure(figsize=(10 , 5))

plt.stackplot(
  month,
  hr,
  marketing,
  sales,
  it,
  labels=[
    "HR",
    "Marketing",
    "Sales",
    "IT"
  ],
  colors=[
    "lightblue",
    "pink",
    "grey",
    "lavender"
  ]
)

plt.xlabel("Month")
plt.ylabel("Revenue")

plt.legend()

plt.title("Department Revenue Distribution")

plt.show()
```

```python
import pandas as pd
import numpy as np
```

#### Reading and Writing CSV , JSON and Excel files

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "age": [24, 30, 22, 35, 28],
    "salary" : [30000 , 45000 , 40000 , 50000 , 25000]
}

df = pd.DataFrame(data)

print(df)
```

```python
import tempfile
from pathlib import Path

csv_path = Path(tempfile.gettempdir()) / "sample.csv"
df.to_csv(csv_path, index=False)

csv_df = pd.read_csv(csv_path)

print(csv_df)
```
